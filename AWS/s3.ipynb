### Making a presigned url for upload

sig version is just the set of rules used to sign, v4 is the standard now.

- for type descriptions to work nicely, you need `boto3-stubs[s3]` as a development dependency
- how does auth on boto work?
    - it didn't ask me anything
    - the way it works is that it tries to look for the same aws cli profile which was configured already by me so wasn't needed again

```python
import boto3
from botocore.config import Config

s3 = boto3.client(
    "s3",
    region_name="us-east-1",
    config=Config(signature_version="s3v4"),
)
```

Now you make the url

```python
Bucket = "ruins-test-bucket2"
url = s3.generate_presigned_url(
    "put_object",
    # can be "put_object" or "get_object"
    # only the corresponding ops are then allowed on the url
    # there is a post as well that we cover later
    Params={
        "Bucket": Bucket,
        "Key": "hello.txt",
    },
    ExpiresIn=300, # seconds
)

url
```

now a simple put

```python
import requests

response = requests.put(url, data="hello from presigned url upload")
response.status_code
```
